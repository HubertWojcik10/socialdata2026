## Part 2: Merging Historical and Recent Crime Data

So far we've been working with crime data from 2018 onwards (I will assume you have that file, named something like "Police Department Incident Reports 2018 to Present"). But San Francisco has published crime data going all the way back to 2003! Having access to this longer time series would let us:
- Study long-term trends in crime
- See how crime patterns changed before, during, and after major events
- Have more statistical power for our analyses

The catch? **The two datasets don't have the same structure.** This is an incredibly common situation in data science. Organizations change how they record data over time, different departments use different formats, and merging datasets requires careful thought.

### Why This Matters

In the real world, you'll almost never get a single, clean dataset that answers your question. Instead, you'll need to combine data from multiple sources: different time periods, different departments, different organizations. Each source will have its own quirks, naming conventions, and missing values.

This is also directly relevant to algorithmic accountability. Remember the Richardson et al. reading from Week 1? They argued that "dirty data" undermines predictive policing. But what counts as "dirty"? If crime categories change over time—if `DRUNKENNESS` disappears as a category in 2018—does that mean public intoxication stopped? Or just that we can't track it anymore? These are the kinds of questions you need to ask whenever you merge datasets.

The skills you'll practice here—schema mapping, category matching, validation—are foundational for any serious data work.

```python
cols_to_keep = [
    "Row ID",
    "Incident Number",
    "Incident Category",
    "Incident Day of Week",
    "Incident Date",
    "Incident Datetime",
    "Incident Time",
    "Police District",
    "Longitude",
    "Latitude",
    "Point",
]

def create_incident_datetime(df):
    df['Incident Datetime'] = pd.to_datetime(
        df['Incident Date'] + ' ' + df['Incident Time'],
        format='%m/%d/%Y %H:%M'
    )
    return df

def modify_incident_datetime(df):
    df['Incident Datetime'] = pd.to_datetime(
        df['Incident Datetime'],
        format='%Y/%m/%d %I:%M:%S %p'
    )
    return df


df_old = create_incident_datetime(df_old_raw)
df_rec = modify_incident_datetime(df_rec_raw)

df_old = df_old[cols_to_keep]
df_rec = df_rec[cols_to_keep]

display(df_old.head())
display(df_rec.head())
```

### Intermezzo: The many flavors of category mismatch

Here are some concrete problems you'll discover:

**Problem 1: Case and formatting differences**

The historical data uses ALL CAPS (`ASSAULT`, `BURGLARY`) while the recent data uses Title Case (`Assault`, `Burglary`). This is easy to fix with `.str.upper()` or `.str.lower()`.

**Problem 2: Name changes**

Some categories look like they were simply renamed:
- `VEHICLE THEFT` → `Motor Vehicle Theft`

**Problem 3: Categories that split or merged**

This is trickier. For example:
- The historical `SEX OFFENSES` might correspond to multiple recent categories: `Sex Offense`, `Rape`, `Human Trafficking`
- The recent data has both `Drug Offense` and `Drug Violation` — should these both map to the historical `DRUG/NARCOTIC`?
- `VANDALISM` in historical data — is that `Vandalism` or `Malicious Mischief` in the recent data? Or both?

**Problem 4: Categories that disappeared or appeared**
- `DRUNKENNESS` exists in historical data but not in recent — was it merged into something else? Decriminalized? Maybe Google it or ask an LLM.
- `Traffic Collision` exists in recent data but not historical — is this a new category, or was it previously recorded differently?
- `Human Trafficking` appears in recent data — is this genuinely new, or was it previously categorized under something else?

**Problem 5: Ambiguous mappings**

Sometimes it's genuinely unclear how categories should map. Does historical `WEAPON LAWS` correspond to recent `Weapons Offense`, `Weapons Carrying Etc`, or both? Without detailed documentation (which often doesn't exist), you have to make judgment calls.

Look at both lists carefully and create mappings based on your judgment.

#### The crucial last step: Data-driven validation

Even after creating a mapping, you should validate it:
- For categories you've matched, do the counts make sense? If `VANDALISM` mapped to `Malicious Mischief`, do you see roughly similar numbers of incidents per year?
- Look at the transition year (2018) — if your mapping is correct, the yearly/monthly trend should make sense. A sudden jump or drop suggests a mapping problem.

```python
category_mapping_old_to_new = {
    # --- High confidence, clean matches ---
    'ASSAULT':                    'Assault',
    'BURGLARY':                   'Burglary',
    'ROBBERY':                    'Robbery',
    'FRAUD':                      'Fraud',
    'ARSON':                      'Arson',
    'EMBEZZLEMENT':               'Embezzlement',
    'DISORDERLY CONDUCT':         'Disorderly Conduct',
    'MISSING PERSON':             'Missing Person',
    'LIQUOR LAWS':                'Liquor Laws',
    'STOLEN PROPERTY':            'Stolen Property',
    'PROSTITUTION':               'Prostitution',
    'SUICIDE':                    'Suicide',
    'GAMBLING':                   'Gambling',
    'RECOVERED VEHICLE':          'Recovered Vehicle',
    'LARCENY/THEFT':              'Larceny Theft',
    'WARRANTS':                   'Warrant',
    'NON-CRIMINAL':               'Non-Criminal',
    'SUSPICIOUS OCC':             'Suspicious Occ',
    'FORGERY/COUNTERFEITING':     'Forgery And Counterfeiting',
    'VEHICLE THEFT':              'Motor Vehicle Theft',

    # --- Medium confidence: renamed or slightly reframed ---
    'VANDALISM':                  'Malicious Mischief',   # SFPD renamed this
    'DRUG/NARCOTIC':              'Drug Offense',         # also 'Drug Violation' exists — see note
    'WEAPON LAWS':                'Weapons Offense',      # also 'Weapons Carrying Etc', 'Weapons Offence'
    'OTHER OFFENSES':             'Other Offenses',
    'DRIVING UNDER THE INFLUENCE':'Traffic Violation Arrest',  # DUI now likely under traffic
    'TRESPASS':                   'Other Offenses',       # no direct match, closest bucket

    # --- Ambiguous / needs validation ---
    'SEX OFFENSES, FORCIBLE':     'Sex Offense',          # could also be 'Rape' — split in new data
    'SEX OFFENSES, NON FORCIBLE': 'Sex Offense',          # same issue
    'KIDNAPPING':                 'Offences Against The Family And Children',  # rough match

    # --- No clear match in recent data ---
    'DRUNKENNESS':                'Other Offenses',       # likely absorbed, no direct equivalent
    'LOITERING':                  'Civil Sidewalks',      # SF-specific reframing post-2018
    'SECONDARY CODES':            None,                   # admin/internal code, drop or ignore
    'EXTORTION':                  'Other Offenses',       # no dedicated category in new data
    'BAD CHECKS':                 'Fraud',                # logical merge
    'BRIBERY':                    'Other Offenses',       # no dedicated category
    'PORNOGRAPHY/OBSCENE MAT':    'Other Offenses',       # no dedicated category
    'TREA':                       None,                   # truncated — likely 'TRESPASS' duplicate?
}

df_old["Incident Category"] = df_old["Incident Category"].map(category_mapping_old_to_new)

# concatenate the two datasets
df_merged = pd.concat([df_old, df_rec])
```

```python
high_confidence_crimes = [
    'Assault',
    'Burglary',
    'Robbery',
    'Fraud',
    'Arson',
    'Embezzlement',
    'Disorderly Conduct',
    'Missing Person',
    'Liquor Laws',
    'Stolen Property',
    'Prostitution',
    'Suicide',
    'Gambling',
    'Recovered Vehicle',
    'Larceny Theft',
    'Warrant',
    'Non-Criminal',
    'Suspicious Occ',
    'Forgery And Counterfeiting',
    'Motor Vehicle Theft'
]

def filter_high_confidence_crimes(df):
    return df[df["Incident Category"].isin(high_confidence_crimes)]
```

```python
import matplotlib.pyplot as plt
```

```python
df_merged["Incident Year"] = df_merged["Incident Datetime"].dt.year
# CASE 1: Yearly trends
fig, axes = plt.subplots(5, 4, figsize=(20, 20))

for i, crime in enumerate(high_confidence_crimes):
    crime_data = df_merged[df_merged['Incident Category'] == crime]
    crime_data.groupby('Incident Year').size().plot(kind='bar', ax=axes[i//4, i%4], width=0.8)
    axes[i//4, i%4].set_title(crime)
    axes[i//4, i%4].set_xlabel('Year')

plt.tight_layout()
```

```python
# CASE 2: Day of the week
fig, axes = plt.subplots(5, 4, figsize=(20, 20))

for i, crime in enumerate(high_confidence_crimes):
    crime_data = df_merged[df_merged['Incident Category'] == crime]
    crime_data.groupby('Incident Day of Week').size().plot(kind='bar', ax=axes[i//4, i%4], width=0.8)
    axes[i//4, i%4].set_title(crime)
    axes[i//4, i%4].set_xlabel('Day of the week')

plt.tight_layout()
```

```python
# CASE 3: 24 hour cycle
# Extract hour from Incident Datetime
df_merged['Incident Hour'] = df_merged['Incident Datetime'].dt.hour

fig, axes = plt.subplots(5, 4, figsize=(20, 20))

for i, crime in enumerate(high_confidence_crimes):
    crime_data = df_merged[df_merged['Incident Category'] == crime]
    crime_data.groupby('Incident Hour').size().plot(kind='bar', ax=axes[i//4, i%4], width=0.8)
    axes[i//4, i%4].set_title(crime)
    axes[i//4, i%4].set_xlabel('Hour of day (0 = midnight)')

plt.tight_layout()
```

```python
# CASE 4: 168 hours of the week
# Create an ordered "hour of week" column: Monday 0am = 0, Monday 1am = 1, ... Sunday 11pm = 167
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

df_merged['Day of Week Num'] = df_merged['Incident Day of Week'].map({day: i for i, day in enumerate(day_order)})
df_merged['Hour of Week'] = df_merged['Day of Week Num'] * 24 + df_merged['Incident Hour']

fig, axes = plt.subplots(5, 4, figsize=(20, 20))

for i, crime in enumerate(high_confidence_crimes):
    crime_data = df_merged[df_merged['Incident Category'] == crime]
    hourly = crime_data.groupby('Hour of Week').size().reindex(range(168), fill_value=0)
    hourly.plot(ax=axes[i//4, i%4], linewidth=0.8)
    
    # Add vertical lines to delineate days
    for day_start in range(0, 168, 24):
        axes[i//4, i%4].axvline(x=day_start, color='gray', linestyle='--', linewidth=0.5, alpha=0.5)
    
    # Label days on x-axis
    axes[i//4, i%4].set_xticks(range(12, 168, 24))
    axes[i//4, i%4].set_xticklabels([d[:3] for d in day_order], rotation=0)
    axes[i//4, i%4].set_title(crime)
    axes[i//4, i%4].set_xlabel('Day of week')

plt.tight_layout()
```

```python
import numpy as np
```

```python
assault = df_merged[df_merged['Incident Category'] == 'Assault']
hourly = assault.groupby('Incident Hour').size()

# Polar bar setup
theta = np.linspace(0, 2 * np.pi, 24, endpoint=False)
width = 2 * np.pi / 24

fig, ax = plt.subplots(figsize=(10, 10), subplot_kw={'projection': 'polar'})
bars = ax.bar(theta, hourly.values, width=width, bottom=0, 
              color=plt.cm.RdYlGn_r(hourly.values / hourly.max()), alpha=0.85)

# Label hours
ax.set_xticks(theta)
ax.set_xticklabels([f'{h:02d}:00' for h in range(24)], fontsize=9)
ax.set_theta_direction(-1)       # clockwise
ax.set_theta_offset(np.pi / 2)  # start at top (midnight)
ax.set_title('Assault by Hour of Day', fontsize=14, pad=20)
plt.show()
```

```python
crimes_to_plot = ['Larceny Theft', 'Assault', 'Burglary']

fig, ax = plt.subplots(figsize=(20, 6))

for crime in crimes_to_plot:
    crime_data = df_merged[df_merged['Incident Category'] == crime]
    monthly = crime_data.set_index('Incident Datetime').resample('ME').size()
    monthly.plot(ax=ax, label=crime)

# Annotate COVID
ax.axvspan('2020-03-01', '2020-06-01', alpha=0.15, color='red', label='COVID lockdown')
ax.set_title('Monthly crime trends in San Francisco (2003–2026)')
ax.set_xlabel('')
ax.set_ylabel('Number of incidents')
ax.legend()
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns

drug = df_merged[df_merged['Incident Category'] == 'Larceny Theft']

day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

# Swap the groupby order: day first, hour second — then unstack puts hours as columns
pivot = drug.groupby(['Incident Day of Week', 'Incident Hour']).size().unstack()
pivot = pivot.reindex(day_order)  # reorder rows instead

fig, ax = plt.subplots(figsize=(12, 8))
sns.heatmap(pivot, cmap='YlOrRd', linewidths=0.3, ax=ax,
            cbar_kws={'label': 'Number of incidents'})
ax.set_title('Drug Offenses: Hour of Day vs Day of Week')
ax.set_xlabel('Hour of day (0 = midnight)')
ax.set_ylabel('')
plt.tight_layout()
plt.show()
```
